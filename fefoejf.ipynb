```python

import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

sample = pd.DataFrame({
    'RM': [6, 7],
    'LSTAT': [12, 10]
})

poly_demo = PolynomialFeatures(degree=2)

result = poly_demo.fit_transform(sample)

new_columns = poly_demo.get_feature_names_out()

pd.DataFrame(result, columns=new_columns)
```
