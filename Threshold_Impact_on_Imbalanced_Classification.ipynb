```python

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_score, recall_score,
    f1_score, average_precision_score
)

# 1. Generate imbalanced data
X, y = make_classification(
    n_samples=2000,
    n_features=10,
    n_informative=4,
    n_redundant=2,
    weights=[0.95, 0.05],
    random_state=42
)

# 2. Stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# 3. Train Logistic Regression
model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

model.fit(X_train, y_train)

# 4. Predicted probabilities
probs = model.predict_proba(X_test)[:, 1]

# 5. Compare fixed thresholds
for threshold in [0.5, 0.3, 0.1]:
    predictions = (probs >= threshold).astype(int)

    print(f"\nThreshold: {threshold}")
    print("Precision:", round(
        precision_score(y_test, predictions, zero_division=0), 3
    ))
    print("Recall:", round(
        recall_score(y_test, predictions), 3
    ))
    print("F1:", round(
        f1_score(y_test, predictions), 3
    ))

print("\nAverage Precision:",
      round(average_precision_score(y_test, probs), 3))
```
